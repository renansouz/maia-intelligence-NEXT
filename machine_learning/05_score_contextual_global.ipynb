# 03 — Saúde Contextual / Global
Segunda camada: combina score operacional + contexto. Pesos são regras explicáveis do MVP.


```python
from pathlib import Path
import sys

# Procura a raiz do projeto
ROOT = Path.cwd().resolve()

for p in [ROOT, *ROOT.parents]:
    if (p / "machine_learning").exists() and (p / "ml_assets").exists():
        ROOT = p
        break

# Adiciona a raiz do projeto aos imports
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from machine_learning.contextual_model import ContextualHealthModel

# Caminhos das configurações
WEIGHTS_PATH = ROOT / "machine_learning" / "config" / "weights.json"
FEEDBACK_PATH = ROOT / "machine_learning" / "config" / "operator_feedback.json"

# Inicializa o modelo contextual
m = ContextualHealthModel(
    WEIGHTS_PATH,
    FEEDBACK_PATH
)

print("Raiz do projeto:", ROOT)
print("Modelo contextual carregado com sucesso.")
```

## Ambiente de teste
`operational_score` vem do Modelo 1. Altere cadastro, ambiente e feedback abaixo.


```python
TESTE={'operational_score':82,'machine_age_years':8,'days_since_maintenance':210,'recommended_maintenance_days':180,'ambient_temp_c':36,'recent_humidity_pct':75,'operator_feedback':[{'category':'noise','severity':'mild'}],'mechanical_claim_severity':0.0}
m.score(**TESTE)
```

## Feedback disponível
Ruído, fumaça, perda de potência, aquecimento percebido, vibração, vazamento, dificuldade de movimentação e alerta no painel. Cada um aceita `mild`, `moderate` ou `severe`. Ajustes ficam em `config/operator_feedback.json`.
